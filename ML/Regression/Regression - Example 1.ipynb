```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
url = "/Workspace/Users/arthur_torres_camargo@hotmail.com/Machine-Learning/ML/Dataset/Diabetes/diabetes_clean.csv"
```

```python
df = pd.read_csv(url)
df.head()
```

```python
## Eliminando os valores 0 do df
df = df[df["glucose"] != 0]
df = df[df["bmi"] != 0]
df.shape
```

## Criação das matrizes de variaveis independentes e dependentes

- Variavel independente: Eliminamos a variavel dependente do conjunto de dados

- Variavel dependente: Usamos só ela com .values

```python
## Variavel independente
X = df.drop("glucose", axis=1).values

## Variavel dependente
y = df["glucose"].values

print(type(X), type(y))
```

## Previsões com base em uma única variavel independente 

```python
X_bmi = X[:, 4]
print(y.shape, X_bmi.shape)
```

```python
## Precisamos transformar nossa matriz da variavel independente em bidimensional
## O -1 indica que o número de linhas será calculado automaticamente com base no tamanho do array original.
## O 1 indica que haverá uma única coluna. 
X_bmi = X_bmi.reshape(-1, 1)
print(y.shape, X_bmi.shape)
```

## Gráfico com os níveis de glicose x BMI

```python
plt.scatter(X_bmi, y)
plt.ylabel("Glicose no sangue (mg/dl)")
plt.xlabel("IMC")
plt.show()
```

## Ajuste de um modelo de regressão

```python
model = LinearRegression()
model.fit(X_bmi, y)

prediction = model.predict(X_bmi)
plt.scatter(X_bmi, y)
plt.plot(X_bmi, prediction, color="red")
plt.ylabel("Glicose no sangue (mg/dl)")
plt.xlabel("IMC")
plt.show()
```

## Como funciona a regressão linear

### y = ax + b
- A regressão linear simples usa uma única variável independente 
  - y = váriavel dependente 
  - x = apenas uma variavel independente 
  - a,b = paramentros/coeficientes do modelo - inclinacao e intercpto
- Como escolher a e b? 
  - Definir uma função de erro para uma detereminada linha
  - Escolher a linha que minimiza a função de erro
- Função de erro = função de perda = função de custo


![Metodo dos minimos quadrados](/Workspace/Users/arthur_torres_camargo@hotmail.com/Machine-Learning/ML/Images/metodo_minimo_quadrados.png)

## Regressão linear com mais dimensões (Regressão Linear Multipla)

### y = a1x1 + a2x2 + b

- Para ajustar um modelo de regressão linear aqui: 
  - É preciso especificar 3 variáveis a1, a2 e b
- Com mais dimensões:
  - É preciso especificar os coeficientes de cada variável independente e a variável b
    - y = a1x1 + a2x2 + a3x3 + ... + anxb + b

```python
## Usando a regressão linear com mais dimensões para diabetes
X_train, X_test,  y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

## R-quadrado

- R²:  Quantifica a variância nos valores da variável dependente explicada pelas variáveis independentes  
  - Os valores variam de 0 a 1 
  - Quando o valor assume 1, significa que todas as variaveis independetes explicam totalmente a variancia da variavel dependente 
  
![R2 ALTO](/Workspace/Users/arthur_torres_camargo@hotmail.com/Machine-Learning/ML/Images/R2 alto e biaxo.png)

### Para calcular o R²

```python
model.score(X_test, y_test)
```

## Erro quadrático médio e raiz do erro quadrático médio

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(Y_i - \hat{Y}_i)^2
$$

- O MSE é uma medida em unidades da variável dependente ao quadrado

- RMSE = &radic;MSE 
  - Meça o RMSE nas mesmas unidades da variável dependente 


```python
## RMSE no scikit-leran
from sklearn.metrics import root_mean_squared_error
```

```python
root_mean_squared_error(y_test, y_pred)
```

## Validação cruzada

- Consiste em criar N grupos para fazer os testes, como na imagem a seguir:

![Validacoes cruzads](/Workspace/Users/arthur_torres_camargo@hotmail.com/Machine-Learning/ML/Images/Sem título.png)

- 5 grupos = validação cruzada com 5 grupos
- 10 grupos = validação cruzada com 10 grupos
- k grupos = validação cruzada com k grupos (k-fold cv)
- Mais grupos = mais caro do ponto de vista computacional

```python
## Validação cruzada
from sklearn.model_selection import cross_val_score, KFold

kf = KFold(n_splits= 6, shuffle=True, random_state=42)
model = LinearRegression()
cv_results = cross_val_score(model, X, y, cv=kf)
```

```python
## Imprimindo resultados da validação cruzada
print(cv_results)
```

```python
## Média e desvio padrão
print(np.mean(cv_results), np.std(cv_results))
```

```python
## Intervalo de confianca de 95%
print(np.quantile(cv_results, [0.025, 0.975]))
```

## Regularização na Regressão

- Por que regularizar? 
  - Lembre-se: A regressão linear minimiza uma função de perda 
  - Escolhe um coeficiente "a", para cada variavel independente, mais "b"
  - Coeficientes grandes podem levar a um sobreajuste 
  - Regularização: Penalização de coeficientes grandes

### Regressão ridge

- Função de perda = função dos minimos quadrados 
- O ridge penaliza coeficientes grandes, positivos ou negativos 
- alfa: parametro que precisamos escolher
- A escolha de alfa no ridge, é semelhante a escolhad e K no KNN
- Hiperparametro: variavel usada para otimizar os parametros do modelo (alfa, nessa caso)
- alfa controla a complexidade do modelo
  - alfa = 0 = MMQ (pode levar a overfitting)
  - alfa muito alto = pode levar underfitting

```python
## Regressão ridge
from sklearn.linear_model import Ridge 
scores = []
for alpha in [0.1, 1.0, 10.0, 100.0, 1000.0]:
    ridge = Ridge(alpha=alpha)
    ridge.fit(X_train, y_train)
    y_pred = ridge.predict(X_test)
    scores.append(ridge.score(X_test, y_test))
```

```python
print(scores)
```

### Regressão Lasso
- Função de perda = função de perda MMQ+ 

```python
## Regressão linear Lasso
from sklearn.linear_model import Lasso
scores = []
for alpha in [0.1, 1.0, 10.0, 20.0, 50.0]:
    lasso = Lasso(alpha=alpha)
    lasso.fit(X_train, y_train)
    y_pred = lasso.predict(X_test)
    scores.append(lasso.score(X_test, y_test))
```

```python
print(scores)
```

### Regressão Linear Lasso para seleção de variáveis independentes
- O lasso pode selecionar variáveis independentes importantes de um conjunto de dados
- Reduz a zero os coeficientes de variáveis independentes menos importantes
- As variaveis independentes não reduzidas a zero são selecionadas pelo lasso

```python
## Lasso para seleção de variaveis independentes 
from sklearn.linear_model import Lasso
X = df.drop("glucose", axis=1).values
y = df["glucose"].values
names = df.drop("glucose", axis=1).columns

lasso = Lasso(alpha=0.1)

model_coef = lasso.fit(X, y).coef_

plt.bar(names, model_coef)
plt.xticks(rotation=45)
plt.show()
```
