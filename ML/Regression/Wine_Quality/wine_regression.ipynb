- Entendemos qual é nossa variavel dependente (quality)

```python
X = df_red.drop('quality', axis=1)
y = df_red['quality']
```

```python
corr_matrix = df_red.corr()

plt.figure(figsize=(12,8))
sns.heatmap(
    corr_matrix,
    cmap='coolwarm',
    annot=True,
    fmt='.2f'
)
```

```python
print(corr_matrix['quality'].sort_values(ascending=False))
```

- Analisando as informações de correlacao: 
    - Guia de Interpretação de Correlação (Variável vs. Target)

| Força da Correlação | Significado | Ação Recomendada |
| :--- | :--- | :--- |
| **0.00 a 0.10** (ou -0.10) | Ruído / Desprezível | **Candidata a descarte.** O Lasso provavelmente zerará seu peso. |
| **0.10 a 0.50** (ou -0.50) | Moderada | **Mantenha.** Ajuda o modelo a entender nuances e detalhes. |
| **Acima de 0.60** (ou -0.60) | Forte | **Essencial.** São as variáveis que ditam a performance do modelo. |

```python
## Eliminando variavies que não apresentam peso
## corr < 0.1 e corr > -0.1
X = X.drop(['pH', 'free sulfur dioxide', 'residual sugar'], axis=1)
```

```python
 print(y.shape, X.shape)
```

MSE (Erro Quadrático Médio): Se ele der, por exemplo, 0.40, lembre-se que esse valor está "ao quadrado". Ele é ótimo para comparar modelos: se o próximo modelo der 0.35, ele é melhor.

RMSE (Raiz do Erro Quadrático Médio): Se o seu RMSE der 0.65, a interpretação para o seu chefe no trabalho seria: "Nosso modelo erra, em média, 0.65 pontos na nota da qualidade do vinho".

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

R^2

```python
model = LinearRegression()
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
score = model.score(X_test, y_test)
print(score)
```

RMSE

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_pred)
print(mse)
```

MSE

```python
from sklearn.metrics import root_mean_squared_error

root_mean_squared_error(y_test, y_pred)
```

Como conferir se o erro está "bom"?



```python
print(f"MSE: {mse}")
print(f"RMSE: {root_mean_squared_error(y_test, y_pred)}")
print(f"Desvio padrão: {df_red['quality'].std()}")
print(f"Score: {model.score(X_test, y_test)}")
```
