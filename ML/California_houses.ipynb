## Descritivo do DF
1. **longitude**: Mede o quão a oeste uma casa está. No contexto da Califórnia, valores mais altos (menos negativos) indicam locais mais ao oeste (próximos à costa).

2. **latitude**: Mede o quão ao norte uma casa está. Valores maiores indicam que a casa fica mais ao norte do estado.

3. **housingMedianAge** (Idade Média das Casas): É a idade mediana das casas naquele bloco. Números baixos significam que as construções são mais novas.

4. **totalRooms** (Total de Cômodos): A soma de todos os cômodos de todas as casas daquele bloco.

5. **totalBedrooms** (Total de Quartos): A soma de todos os quartos de dormir das casas daquele bloco.

6. **population** (População): O número total de pessoas que moram naquele bloco.

7. **households** (Famílias/Domicílios): O número total de unidades familiares (grupos de pessoas que vivem em uma casa) no bloco.

8. **medianIncome** (Renda Mediana): A renda média das famílias daquele bloco. Atenção: Está em dezenas de milhares de dólares (ex: se o valor for 3.0, significa US$ 30.000,00 por ano).

9. **medianHouseValue** (Valor Mediano da Casa): O valor médio das casas naquele bloco em dólares americanos. Geralmente, este é o "alvo" que queremos prever em modelos de inteligência artificial.

10. **oceanProximity** (Proximidade do Oceano): Indica onde a casa está em relação ao mar (ex: "Perto do mar", "No interior", "Numa ilha").
