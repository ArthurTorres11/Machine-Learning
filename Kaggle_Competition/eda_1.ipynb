```python
 ## Como só temos duas linhas com valores nulos, podemos exlcuir
train_df.dropna(subset=["Embarked"], inplace=True)
print(train_df["Embarked"].isnull().sum())
```

```python
## Usando o One Hot Enconder para a coluna Embarked
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder()
train_encoder = encoder.fit_transform(train_df[["Embarked"]])
```

```python
train_encoder = train_encoder.toarray()
```

```python
## Transformando o array em Dataframe
df_encoded = pd.DataFrame(train_encoder)
df_encoded.head()
```

```python
train_df = pd.concat([train_df, df_encoded], axis=1)
```

```python
train_df.head()
```

```python
train_df.drop(columns=['Embarked'], inplace=True)
```

```python
train_df.rename(columns= {
    0: "Embarked_C",
    1: "Embarked_Q",
    2: "Embarked_S"
},
                inplace=True)
```

```python
train_df.head()
```

```python
## SibSp + Parch
## Entedemos que as duas falam sobre a quantidade de familiares que o passageiro levava consigo
## Podemos criar uma coluna que some as duas e assim teremos uma coluna que fala sobre a quantidade de familiares que o passageiro levava consigo

train_df["qtd_familia"] = train_df["SibSp"] + train_df["Parch"] 
train_df.drop(columns=["SibSp", "Parch"], inplace=True)
train_df.head()
```

```python
## Transformando agora para uma coluna binária pensando na melhora do modelo
train_df["viajava_sozinho"] = train_df["qtd_familia"].apply(lambda x: 1 if x == 0 else 0)
train_df.drop(columns=["qtd_familia"], inplace=True)
```

### Vamos seguir para o tratamento da idade

```python
print(f"Valores faltantes da coluna Age: {train_df["Age"].isna().sum()}")
print(f"Valor totais da coluna Age: {train_df["Age"].count()}")
```

```python
## Como podemos tratar essa idaide, vamos analisar a idade média por classe e fazer as verificações
train_df.groupby("Pclass")["Age"].median()
```

```python
train_df["Age"] = train_df["Age"].fillna(train_df.groupby("Pclass")["Age"].transform("median"))
train_df.head()
print(f"Valores nulos: {train_df["Age"].isna().sum()}")
```
